```python
import requests
from rdflib import Graph
from rdflib.namespace import SKOS, RDF, DCTERMS
import json
import logging
logging.getLogger("rdflib.term").setLevel(logging.ERROR)
import Function_Data as f
```

```python
THESO_ID = "th1200"
url = f"https://opentheso2.mom.fr/api/all/theso?id={THESO_ID}"
```

```python
resp = requests.get(url, headers={"Accept": "application/rdf+xml"})
resp.raise_for_status()
```

```python
g = Graph()
g.parse(data=resp.text, format="xml")
```

```python
print(f"{len(g)} triplets récupérés")
```

```python
for sujet, predicat, objet in g:
    print(f"{sujet}  |  {predicat}  |  {objet}")
```

```python
for theso, titre in g.subject_objects(DCTERMS.title):
    print(f"Thésaurus : {titre}\n")
```

```python
for concept in g.subjects(RDF.type, SKOS.Concept):
    labels = list(g.objects(concept, SKOS.prefLabel))
    label = labels[0] if labels else "(sans label)"
    print(f"{label}")
```

```python
print("=" * 60)
print("THÉSAURUS")
print("=" * 60)
for scheme in g.subjects(RDF.type, SKOS.ConceptScheme):
    print(f"\nURI : {scheme}")
    for predicat, objet in g.predicate_objects(scheme):
        nom_predicat = g.namespace_manager.normalizeUri(predicat)
        print(f"   {nom_predicat} : {objet}")
```

```python
print("\n" + "=" * 60)
print("CONCEPTS")
print("=" * 60)
for concept in g.subjects(RDF.type, SKOS.Concept):
    print(f"\nURI : {concept}")
    for predicat, objet in g.predicate_objects(concept):
        nom_predicat = g.namespace_manager.normalizeUri(predicat)
        print(f"   {nom_predicat} : {objet}")
```

```python
print("=" * 60)
print("CONCEPTS")
print("=" * 60)
for concept in g.subjects(RDF.type, SKOS.Concept):
    print(f"\n {f.label(g, concept)}  (id: {concept})")

    for parent in g.objects(concept, SKOS.broader):
        print(f"   broader  (parent)  : {f.label(g, parent)}")

    for enfant in g.objects(concept, SKOS.narrower):
        print(f"   narrower (enfant)  : {f.label(g, enfant)}")

    if list(g.objects(concept, SKOS.topConceptOf)):
        print(f"   → concept racine")
```

```python
concepts = []
for c in g.subjects(RDF.type, SKOS.Concept):
    concepts.append({
        "id": str(g.value(c, DCTERMS.identifier) or c),
        "uri": str(c),
        "prefLabel": f.label(g, c),
        "definition": [str(d) for d in g.objects(c, SKOS.definition)],
        "broader": [f.label(g, p) for p in g.objects(c, SKOS.broader)],
        "narrower": [f.label(g, e) for e in g.objects(c, SKOS.narrower)],
        "isTopConcept": bool(list(g.objects(c, SKOS.topConceptOf))),
        "created": str(g.value(c, DCTERMS.created) or ""),
        "modified": str(g.value(c, DCTERMS.modified) or ""),
    })

print(json.dumps(concepts, ensure_ascii=False, indent=2))
print(f"\n{len(concepts)} concepts")
```

```python
BASE = "https://opentheso2.mom.fr"
            
r = requests.get(f"{BASE}/openapi/v1/thesaurus",headers={"Accept": "application/json"})

print("Status:", r.status_code)
print(r.text[:2000]) 
```

```python
thesos = requests.get(f"{BASE}/openapi/v1/thesaurus",headers={"Accept": "application/json"}).json()
print(f"{len(thesos)} thésaurus à traiter\n")
```

```python
resultat = {}
for t in thesos:
    tid = t["idTheso"]
    titre = f.titre_fr(t)
    try:
        resp = requests.get(f"{BASE}/api/all/theso?id={tid}",headers={"Accept": "application/rdf+xml"}, timeout=120)
        g = Graph()
        g.parse(data=resp.text, format="xml")

        complets = [f.extraire_concept(g, c)for c in g.subjects(RDF.type, SKOS.Concept)]
        complets = [c for c in complets if f.a_attribut_complet(c)]

        resultat[tid] = {"titre": titre, "concepts": complets}
        print(f"{tid:10} | {len(complets):5} concepts complets | {titre}")
    except Exception as e:
        print(f"{tid:10} | ERREUR : {e}")
        resultat[tid] = {"titre": titre, "concepts": [], "erreur": str(e)}

total = sum(len(v["concepts"]) for v in resultat.values())
print(f"\nTOTAL : {total} concepts complets sur {len(thesos)} thésaurus")
```

```python
import os

DOSSIER = "thesaurus_data2"
os.makedirs(DOSSIER, exist_ok=True)
meta_par_id = {t["idTheso"]: t for t in thesos}
```

```python
sauvegardes = 0
total_concepts = 0
for tid, data in resultat.items():
    complets = [f.nettoyer(c) for c in data["concepts"] if f.est_complet(c)]
    if not complets:
        continue
    sortie = {
        "thesaurus": f.meta_thesaurus(tid, meta_par_id),
        "nb_concepts": len(complets),
        "concepts": complets,
    }
    with open(os.path.join(DOSSIER, f"{tid}.json"), "w", encoding="utf-8") as fichier:
        json.dump(sortie, fichier, ensure_ascii=False, indent=2)
    sauvegardes += 1
    total_concepts += len(complets)
    print(f"{tid:10} | {len(complets):5} concepts | {f.meta_thesaurus(tid, meta_par_id)['titres'].get('fr', tid)}")

print(f"\n{sauvegardes} thésaurus sauvegardés | {total_concepts} concepts au total")
```

```python
import csv

ATTRS_PRINCIPAUX = ["prefLabel", "altLabel", "definition", "broader", "narrower"]
```

```python
def est_complet(c):
    return all(c.get(a) for a in ATTRS_PRINCIPAUX)

def joindre(val):
    if isinstance(val, list):
        return " | ".join(str(x) for x in val)
    return str(val) if val is not None else ""

```

```python
FICHIER = "concepts_complets.csv"
colonnes = ["id", "uri", "prefLabel", "altLabel", "definition","broader", "narrower", "isTopConcept"]

```

```python
nb = 0
with open(FICHIER, "w", newline="", encoding="utf-8-sig") as f:
    writer = csv.writer(f)
    writer.writerow(colonnes)                      
    for tid, data in resultat.items():
        for c in data["concepts"]:
            if not est_complet(c):                
                continue
            writer.writerow([
                c["id"],
                c["uri"],
                c["prefLabel"],
                joindre(c["altLabel"]),
                joindre(c["definition"]),
                joindre(c["broader"]),
                joindre(c["narrower"]),
                c["isTopConcept"],
            ])
            nb += 1

print(f"{nb} concepts complets écrits dans {FICHIER}")
```
